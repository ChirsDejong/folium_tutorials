导入folium

```python
import folium
import folium.plugins
```

```python
# 上海坐标
shanghai = [31.35, 121.10]
```

实例化地图

```python
# 创建地图
m = folium.Map(
    location=shanghai,
    zoom_start=4,
    tiles=None,  # 不使用默认底图
)

minimap = folium.plugins.MiniMap(
    toggle_display=True,
    position='bottomleft',
    zoom_level_fixed=-4,
    
    width=200,
    height=200,   
)
```

```python
m.add_child(minimap)

# 添加高德地图底图
folium.TileLayer(
    tiles='http://webrd0{s}.is.autonavi.com/appmaptile?lang=zh_cn&size=1&scale=1&style=8&x={x}&y={y}&z={z}',
    attr='&copy; <a href="https://map.amap.com">高德地图</a>',
    name='高德地图',
    max_zoom=18,
    subdomains='1234',
    overlay=False,
    control=True
).add_to(m)

m
```
