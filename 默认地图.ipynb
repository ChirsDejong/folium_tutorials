导入folium

```python
import folium
from branca.element import Figure
```

```python
# 上海坐标
shanghai = [31.35, 116.10]
```

实例化地图

```python
# 创建地图
TILES = "openstreetmap"
m = folium.Map(
    location=shanghai,
    zoom_start=10,
    tiles=TILES,
)
m
```

```python
# 添加不同的绘图风格选框
m = folium.Map(
    location=[31.35, 116.10],
    zoom_control=True,
    zoom_start=12,
    conrol_scale=True,
    tiles = 'openstreetmap')

fig = Figure(width=750, height=450)
fig.add_child(m)

tile_list = [
    ('openstreetmap', '© OpenStreetMap contributors'),
    ('stamenwatercolor', 'Map tiles by Stamen Design'),
    ('cartodbdark_matter', '© CARTO'),
    ('cartodbpositron', '© CARTO'),
]

for each, attr in tile_list:
    folium.TileLayer(each, attr=attr).add_to(m)
    
folium.LayerControl().add_to(m)

m
```
