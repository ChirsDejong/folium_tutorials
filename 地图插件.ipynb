导入folium

```python
import folium
```

```python
# 定义上海坐标 [纬度, 经度]
shanghai = [9.35, 121.10]
```

实例化地图

```python
# 创建地图
# location: 地图中心点坐标
# zoom_start: 初始缩放级别
# zoom_control: 是否显示缩放控件
# control_scale: 是否显示比例尺
m = folium.Map(
    location=shanghai,
    zoom_start=10,
    zoom_control=True,      # 显示缩放控件
    control_scale=True,     # 显示比例尺
)
```

```python
# 添加高德地图底图
# tiles: 瓦片服务URL模板
# attr: 版权声明
# name: 图层名称
# max_zoom: 最大缩放级别
# subdomains: 子域名轮询(用于负载均衡)
# overlay: 是否作为覆盖层
# control: 是否显示图层控制
folium.TileLayer(
    tiles='http://webrd0{s}.is.autonavi.com/appmaptile?lang=zh_cn&size=1&scale=1&style=8&x={x}&y={y}&z={z}',
    attr='&copy; <a href="https://map.amap.com">高德地图</a>',
    name='高德地图',
    max_zoom=18,
    subdomains='1234',
    overlay=False,
    control=True
).add_to(m)
```

```python
m
```

插件-点击鼠标，添加大头钉

```python
# 添加点击添加标记插件
# popup: 标记点击时显示的弹出内容
plugin_clickpoint = folium.ClickForMarker(popup='point')

m.add_child(plugin_clickpoint)
```

插件-点击鼠标显示经纬度

```python
# 添加点击显示经纬度插件
# 该插件会在点击位置显示一个包含经纬度信息的弹出框
plugin_clickpoint_LatLon = folium.LatLngPopup()

m.add_child(plugin_clickpoint_LatLon)
```

插件-移动鼠标，显示经纬度

```python
from folium.plugins import MousePosition
# 插件-移动鼠标，显示经纬度
formatter ="function(num){return L.Util.formatNum(num, 4)+ ' º ';};"
plugin_hover =MousePosition(
    position='topright',                    # 控件显示在地图右上角
    separator=' ',                          # 经纬度值之间的分隔符为空格
    empty_string='鼠标划动显示经纬度',         # 初始提示文字
    lng_first=False,                        # 显示顺序为纬度在前
    num_digits=20,                          # 显示20位小数精度
    prefix='经纬度:',                        # 坐标值前缀文字
    lat_formatter=formatter,
    lng_formatter=formatter,
)

m.add_child(plugin_hover)
```

插件-禁止缩放地图按钮

```python
# 创建滚动缩放切换器插件实例，允许通过按钮控制是否启用鼠标滚轮缩放功能
plugin_np_zoom = folium.plugins.ScrollZoomToggler()
plugin_np_zoom.add_to(m)

m
```

插件-晨昏线

```python
# 设置初始地图中心点为纬度45°、经度3°，设置初始缩放级别为1（最远视角）
m = folium.Map(location=[45,3], zoom_start=1)

# 创建了一个Terminator插件实例，这个插件会在地图上显示昼夜分界线（晨昏线）
plugin_terminator = folium.plugins.Terminator()
plugin_terminator.add_to(m)

m
```

插件——全屏按钮

```python
# 插件-全屏按钮
# 添加全屏控制插件，允许用户将地图扩展到全屏模式
m = folium.Map(location=[31.35, 116.10], zoom_start=4)

# 插件-全屏按钮
plugin_fullscreen = folium.plugins.Fullscreen(
    position='topright',                # 控件位置设置在地图的右上角
    title='Expand me',                  # 鼠标悬停在按钮上时显示的提示文本
    title_cancel='Exit me',             # 全屏模式下鼠标悬停在按钮上时显示的提示文本
    force_separate_button=True          # 强制创建一个独立的按钮，而不是与其他控件合并
)
# 将全屏控件添加到地图对象中
plugin_fullscreen.add_to(m)

m
```

插件——缩略图

```python
# 创建一个以中国中部为中心的地图，初始缩放级别为4
m = folium.Map(
    location=[31.35, 121.10],
    zoom_start=4
)

minimap = folium.plugins.MiniMap(
    toggle_display=True,
    position='bottomleft',
    zoom_level_fixed=-4,
    
    width=200,
    height=200,   
)

m.add_child(minimap)

m
```

插件——搜索框Geocoder

```python
plugin_geocoder = folium.plugins.Geocoder()
plugin_geocoder.add_to(m)
m
```

插件——绘图工具

```python
plugin_draw = folium.plugins.Draw()
m.add_child(plugin_draw)
m
```

插件——测量工具

```python
plugin_measure = folium.plugins.MeasureControl()
m.add_child(plugin_measure)

m
```

插入自定义图标

```python
from folium.features import CustomIcon

icon_image = 'http://leafletjs.com/examples/custom-icons/leaf-red.png'
shadow_image = 'http://leafletjs.com/examples/custom-icons/leaf-shadow.png'

icon = CustomIcon(
    icon_image,
    icon_size=(38, 95),
    icon_anchor=(22, 94),
    shadow_image=shadow_image,
    shadow_size=(50, 64),
    shadow_anchor=(4, 62),
    popup_anchor=(-3, -76),
)

marker = folium.Marker(
    location=[31.2886, 121.2059],
    icon=icon,
    popup="Golf",
    tooltip='Custom Icon'
)

m.add_child(marker)

m
```

```python
import folium
from folium.plugins import MousePosition, ScrollZoomToggler, Terminator, Fullscreen, MiniMap, Geocoder, Draw, MeasureControl
from folium.features import CustomIcon

# 定义上海坐标 [纬度, 经度]
shanghai = [31.35, 121.10]

# 创建地图
m = folium.Map(
    location=shanghai,
    zoom_start=10,
    zoom_control=True,      # 显示缩放控件
    control_scale=True,     # 显示比例尺
)

# 添加高德地图底图
folium.TileLayer(
    tiles='http://webrd0{s}.is.autonavi.com/appmaptile?lang=zh_cn&size=1&scale=1&style=8&x={x}&y={y}&z={z}',
    attr='&copy; <a href="https://map.amap.com">高德地图</a>',
    name='高德地图',
    max_zoom=18,
    subdomains='1234',
    overlay=False,
    control=True
).add_to(m)

# # 1. 添加点击添加标记插件
# plugin_clickpoint = folium.ClickForMarker(popup='point')
# m.add_child(plugin_clickpoint)

# 2. 添加点击显示经纬度插件
plugin_clickpoint_LatLon = folium.LatLngPopup()
m.add_child(plugin_clickpoint_LatLon)

# 3. 添加鼠标移动显示经纬度插件
formatter = "function(num){return L.Util.formatNum(num, 4)+ ' º ';}"
plugin_hover = MousePosition(
    position='topright',                    # 控件显示在地图右上角
    separator=' ',                          # 经纬度值之间的分隔符为空格
    empty_string='鼠标划动显示经纬度',         # 初始提示文字
    lng_first=False,                        # 显示顺序为纬度在前
    num_digits=20,                          # 显示20位小数精度
    prefix='经纬度:',                        # 坐标值前缀文字
    lat_formatter=formatter,
    lng_formatter=formatter,
)
m.add_child(plugin_hover)

# 4. 添加滚动缩放切换器插件
# plugin_np_zoom = ScrollZoomToggler()
plugin_np_zoom = ScrollZoomToggler(position='bottomright')
m.add_child(plugin_np_zoom)

# # 5. 添加晨昏线插件
# plugin_terminator = Terminator()
# m.add_child(plugin_terminator)

# 6. 添加全屏按钮插件
plugin_fullscreen = Fullscreen(
    position='topright',                # 控件位置设置在地图的右上角
    title='Expand me',                  # 鼠标悬停在按钮上时显示的提示文本
    title_cancel='Exit me',             # 全屏模式下鼠标悬停在按钮上时显示的提示文本
    force_separate_button=True          # 强制创建一个独立的按钮，而不是与其他控件合并
)
m.add_child(plugin_fullscreen)

# 7. 添加缩略图插件
minimap = MiniMap(
    toggle_display=True,
    position='bottomright',
    zoom_level_fixed=-4,
    width=200,
    height=200,   
)
m.add_child(minimap)

# 8. 添加搜索框插件
plugin_geocoder = Geocoder()
m.add_child(plugin_geocoder)

# 9. 添加绘图工具插件
plugin_draw = Draw()
m.add_child(plugin_draw)

# 10. 添加测量工具插件
plugin_measure = MeasureControl()
m.add_child(plugin_measure)

# # 11. 添加自定义图标
# icon_image = 'http://leafletjs.com/examples/custom-icons/leaf-red.png'
# shadow_image = 'http://leafletjs.com/examples/custom-icons/leaf-shadow.png'

# icon = CustomIcon(
#     icon_image,
#     icon_size=(38, 95),
#     icon_anchor=(22, 94),
#     shadow_image=shadow_image,
#     shadow_size=(50, 64),
#     shadow_anchor=(4, 62),
#     popup_anchor=(-3, -76),
# )

# marker = folium.Marker(
#     location=[31.2886, 121.2059],
#     icon=icon,
#     popup="Golf",
#     tooltip='Custom Icon'
# )
# m.add_child(marker)

# 显示地图
m
```
